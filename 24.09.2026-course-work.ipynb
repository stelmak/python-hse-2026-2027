```python
"""
Презентация 2 — повышенный уровень.
Задачи уровня 2 (2.1–2.3) и уровня 3 (3.1, моржовый оператор).

Решения писать здесь. Тесты — в конце файла.
"""

# ============================================================
# Задача 2.1. Время по секундам
# ============================================================
# Дано целое число секунд. Переведите его в формат ЧЧ:ММ:СС
# с помощью арифметических операций (// и %).
#
# Пример:
#   In: 3661
#   Out: 01:01:01
#
# Подсказка: часы = s // 3600; минуты = (s % 3600) // 60;
#            секунды = s % 60.

def seconds_to_hms(seconds):
    hours, minutes, second = seconds // 3600, (seconds % 3600) // 60, seconds % 60
    return f'{hours:02}:{minutes:02}:{second:02}'


# ============================================================
# Задача 2.2. Сумма цифр трёхзначного числа
# ============================================================
# Дано трёхзначное число. Найдите сумму его цифр, используя
# только арифметические операции.
#
# Пример:
#   In: 456
#   Out: 15 (4 + 5 + 6)
#
# Подсказка: n // 100, (n // 10) % 10, n % 10.

def sum_digits(n: int) -> int:
    new_list = [n // 100, (n // 10) % 10, n % 10]
    return sum(new_list)


# ============================================================
# Задача 2.3. Размен монет
# ============================================================
# Дана сумма в копейках. Определите минимальное количество
# монет номиналом 1, 5, 10, 50 копеек, чтобы выдать эту сумму.
#
# Пример:
#   In: 87
#   Out:
#     50: 1
#     10: 3
#     5: 1
#     1: 2
#
# Подсказка: используйте целочисленное деление // и остаток %
# последовательно для каждого номинала.

def coin_change(amount: int) -> dict:
    out = {
      50: amount // 50,
      10: (amount % 50) // 10,
      5: ((amount % 50) % 10) // 5,
      1: (((amount % 50) % 10) % 5) // 1}
    return out
    


# ============================================================
# Задача 3.1. Ввод с проверкой (моржовый оператор)
# ============================================================
# Запросите у пользователя число. Используя моржовый оператор (:=),
# сохраните результат в переменную прямо внутри условия if
# и проверьте, что число положительное.
#
# Пример:
#   In: 5
#   Out: Положительное: 5
#
#   In: -3
#   Out: Не положительное
#
# Подсказка:
#   if (n := int(input())) > 0:
#       print(f"Положительное: {n}")
#   else:
#       print("Не положительное")

def check_positive():
    if (n := int(input())) >= 0:
        print(f'Положительное: {n}')
    else:
        print('Не положительное')


# ============================================================
# Тесты (assert)
# ============================================================

if __name__ == "__main__":
    # 2.1
    assert seconds_to_hms(3661) == "01:01:01"
    assert seconds_to_hms(0) == "00:00:00"
    assert seconds_to_hms(59) == "00:00:59"
    assert seconds_to_hms(3600) == "01:00:00"

    # 2.2
    assert sum_digits(456) == 15
    assert sum_digits(100) == 1
    assert sum_digits(999) == 27

    # 2.3
    assert coin_change(87) == {50: 1, 10: 3, 5: 1, 1: 2}
    assert coin_change(0) == {50: 0, 10: 0, 5: 0, 1: 0}
    assert coin_change(50) == {50: 1, 10: 0, 5: 0, 1: 0}

    # 3.1 — читает с клавиатуры, проверяется вручную
    print("p2_advanced: все тесты пройдены (кроме check_positive)")
```
